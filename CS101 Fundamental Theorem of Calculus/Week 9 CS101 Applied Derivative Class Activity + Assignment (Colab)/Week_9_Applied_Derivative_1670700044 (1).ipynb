```
#import/initialize
from sympy import *
#create symbols
x,y,z = symbols('x y z')
#create expressions
exp1 = (x**2+1)/(x-1)
print("Before Substitution : ",exp1)
#.subs ใช้แทนค่าตัวแปรเพื่อหาผลลัพธ์
print("After Substitution  : ",exp1.subs(x,2))
exp2 = x**2 + 7 * y + z
print("Before Substitution : ",exp2)
print("After Substitution  : ",exp2.subs([(x, 2), (y, 4), (z, 1)]) )
print("----------------")
#diff >> derivative ใช้ในการหาอนุพันธ์
exp3 = x**4
print("f = ",exp3)
print("f'(x)=",diff(exp3)) #หาอนุพันธ์อันดับ 1
print("f''(x)=",diff(exp3,x,2)) #หาอนุพันธ์อันดับ 2
print("f'''(x)=",diff(exp3,x,3)) #หาอนุพันธ์อันดับ 3
print("----------------")
#parse_expr >> From string to sympy expression
exp = parse_expr(input("กรอกสมการฟังก์ชัน f(x) : "))#From string to sympy expression
num = int(input("กรอกจำนวนครั้งที่ต้องการหาอนุพันธ์ : "))
for i in range(1,num+1):
  print("f",end="")
  for j in range(1,i+1):
    print("'",end="")
  print("(x) = ",end="")
  print(diff(exp,x,i))
print("----------------")
#method.solve ใช้ในการหาค่าตัวแปรโดยกำหนดสมการให้เท่ากับ 0
tc = 2*x-160
val_x = solve(tc) #ผลลัพธ์ที่ได้จะอยู่ในชนิดข้อมูลโครงสร้าง list
print(tc,"= 0")
print("x = ",val_x[0])
```

```
from sympy import *
q = symbols('q')
tc = parse_expr(input("กรอกสมการฟังก์ชันต้นทุน: TC(q): "))#From string to sympy expression
#tc = 2*q**2-160*q+5000
#ขั้นตอนที่ 1  หาอนุพันธ์อันดับ 1
dif1_tc = diff(tc)
print(f"TC'(q): = {dif1_tc}")

#ขั้นตอนที่ 2 กำหนดอนุพันธ์อันดับ 1 ให้เท่ากับ 0 เพื่อแก้สมการ
val_q = solve(dif1_tc)
print(f"แก้สมการหาค่า q มีค่าเท่ากับ {val_q[0]}")

#ขั้นตอนที่ 3 ทดสอบโดยการหาอนุพันธ์อันดับ 2
print("ทดสอบค่า q ที่หาได้โดยการหาอุนพันธ์อันดับที่ 2")
dif2_tc = diff(dif1_tc)
#dif2_tc = ต่าง(tc,q,2)

#ขั้นตอนที่ 4 แทนค่า q
val_tc = tc.subs(q,val_q[0])
if dif2_tc > 0:
  print(f"TC' ' (q): = {dif2_tc} มีค่าเป็น + ")
  print(f"ดังนั้นที่การผลิตสินค้า {val_q[0]} หน่วย ทำให้มีต้นทุนต่ำสุด {val_tc} บาท")
elif dif2_tc < 0:
  print(f"TC''(q): = {dif2_tc} มีค่าเป็น - ")
  print(f"ดังนั้นที่การผลิตสินค้า {val_q[0]} หน่วย ไม่ทำให้เกิดต้นทุนต่ำสุด")
else :
  print(f"TC''(q): = 0 ไม่สามารถบอกได้ว่าเป็นค่าต่ำสุดหรือสูงสุด ")
```
