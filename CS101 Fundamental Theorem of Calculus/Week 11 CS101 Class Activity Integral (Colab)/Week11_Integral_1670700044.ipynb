```python
#ปริพันธ์จำกัดเขต
#import/initialize
from sympy import *
#create symbols
x = symbols('x')
exp = parse_expr(input("กรอกสมการ : ")) #exp=x**2-3
expr = integrate(exp)
print("สมการที่อินทิเกรต = ",expr)
a = int(input("กรุณากรอกค่า a :" )) #a=1
b = int(input("กรุณากรอกค่า b :" )) #b=2
#.subs ใช้แทนค่าตัวแปรเพื่อหาผลลัพธ์
#ให้นักศึกษาเขียนคำสั่งเพิ่มเติมเพื่อให้ได้ผลลัพธ์การทำงานตามที่โจทย์กำหนด
expr_a = expr.subs(x,a)
expr_b = expr.subs(x,b)
result = expr_b - expr_a
print(f"({expr_b}) - ({expr_a}) = {result}")
```

```python
#import/initialize
from sympy import *
#create symbols
q = symbols('q')
mc = parse_expr(input("กรอกฟังก์ชันต้นทุนเพิ่ม (MC) : "))
tc = integrate(mc)
fixc = int(input("กรุณากรอกต้นทุนคงที่ : "))
#ให้นักศึกษาเขียนคำสั่งเพิ่มเติมเพื่อให้ได้ผลลัพธ์การทำงานตามที่โจทย์กำหนด
tc = tc + fixc
print(f"ฟังก์ชันต้นทุนรวม TC(q) = {tc}")
val_q = int(input("กรอกจำนวนสินค้าที่ต้องการผลิด (q) : "))
val_tc = tc.subs(q,val_q)
print(f"บริษัทจะมีต้นทุนในการผลิตสินค้า {val_q} หน่วย คิดเป็นเงิน {val_tc:.2f} บาท")
```

```python
#import/initialize*
from sympy import *
#create symbols
q = symbols('q')
mr = parse_expr(input("กรอกฟังก์ชันรายรับเพิ่ม (MR) : "))
#ให้นักศึกษาเขียนคำสั่งเพิ่มเติมเพื่อให้ได้ผลลัพธ์การทำงานตามที่โจทย์กำหนด
tr = integrate(mr)
print(f"รายรับรวม TR(q) = {tr}")
val_q  = int(input("กรุณากรอกจำนวนสินค้าที่ขายได้ (q) : "))
val_tr = tr.subs(q,val_q)
print(f"โรงงานจะมีรายรับจากการขายสินค้า {val_q} หน่วย เป็นเงินจำนวน {val_tr:.2f} บาท")
```

```python
#import/initialize
from sympy import *
#create symbols
q = symbols('q')
mc = parse_expr(input("กรอกฟังก์ชันต้นทุนเพิ่ม (MC) : "))
tc = integrate(mc)
fixc = int(input("กรุณากรอกต้นทุนคงที่ : "))
tc = tc + fixc
print(f"ฟังก์ชันต้นทุน (TC) = {tc}")
mr = parse_expr(input("กรอกฟังก์ชันรายรับเพิ่ม (MR) : "))
tr = integrate(mr)
print(f"ฟังก์ชันรายรับรวม (TR) = {tr}")
#ให้นักศึกษาเขียนคำสั่งเพิ่มเติมเพื่อให้ได้ผลลัพธ์การทำงานตามที่โจทย์กำหนด
p = tr - tc
print(f"ฟังก์ชันกำไร P(q) = {p}")
val_q= int(input("กรอกจํานวนสินค้าที่ผลิตและขายได้ (q): "))
val_p = p.subs(q, val_q)
print (f"โรงงานจะมีกำไรจากการขายสินค้า {val_q} หน่วย เป็นจํานวนเงิน {val_p:.2f} บาท")
```
